# Análisis del balance de materia en fabricación de queso mediante `Python`

```python
import os, sys
import pandas as pd
import datetime
import numpy as np
import matplotlib.pyplot as plt
```

```python
import seaborn as sns
sns.set_style("whitegrid")
```

```python
url_datos = 'https://raw.githubusercontent.com/juanriera/master-queseria/master/datos/fab_queso_bm.csv'

try:
    df = pd.read_csv(url_datos, decimal = ",", sep=';', encoding='ISO-8859-1')
    print("¡Archivo ", url_datos," cargado correctamente!")
    print(f"Dimensiones de los datos: {df.shape}")
except Exception as e:
    print(f"Error al cargar el archivo: {e}")
```

```python
# crear index con fecha
df['fecha_index'] = pd.to_datetime(
    df['fecha'], 
    format='%d/%m/%Y',  # Formato Día/Mes/Año confirmado
    errors='coerce'     # Clave: convierte los valores que no puede leer a NaT (Not a Time)
)
# 2. Limpiar NaT y establecer el índice
df.dropna(subset=['fecha_index'], inplace=True)
df.set_index('fecha_index', inplace=True)
df.sort_index(inplace=True)
```

```python
# Crear una lista de los nombres de recetas únicos en el orden en que aparecen
recetas_unicas_ordenadas = df['receta'].drop_duplicates().tolist()

# Crear un diccionario para mapear cada nombre único a su número secuencial
# Ejemplo: {'IQM pequeño': 1, 'Oveja DO 3 kg': 2, ...}
mapa_secuencial = {nombre: i + 1 for i, nombre in enumerate(recetas_unicas_ordenadas)}

# Asignar el nuevo número secuencial a la nueva columna
df['num_receta'] = df['receta'].map(mapa_secuencial)

# Finalmente, crear el formato de texto 'receta X'
df['receta'] = 'receta ' + df['num_receta'].astype(str)

df.to_csv("fab_queso_bm_recetas_secuenciales.csv", index=False, decimal = ",", sep=';')
```
