## 🏭 Análisis de datos en producción: ficha didáctica

### 🎯 Objetivo
Comprender cómo el análisis de datos en procesos industriales va más allá del registro contable, y cómo herramientas como Python permiten mejorar la trazabilidad, la eficiencia y la comprensión tecnológica de los procesos.

---

### 🧠 1. ¿Por qué analizamos los datos en producción?

- No buscamos solo dejar constancia del resultado, como haría un notario o contable.
- Lo importante es poder revisar los resultados en distintos momentos, detectar tendencias y cruzar variables.
- El análisis es parte de un ciclo de mejora continua: sirve para reducir pérdidas y entender los procesos que las generan.

---

### 🔁 2. Mejora continua y trazabilidad

- El análisis no es un fin, sino un motor de cambio.
- Nos ayuda a identificar qué procesos están generando pérdidas y cómo corregirlos.
- La trazabilidad permite que otros técnicos comprendan, reproduzcan y mejoren el análisis.

---

### 🐍 3. ¿Por qué usar Python?

- Separa claramente los datos (archivo .csv) del análisis (script).
- Permite agrupar por fechas, aplicar filtros y explorar nuevas vistas de los datos.
- Documenta todo el proceso: limpieza, análisis, visualización.

---

### 📊 4. Comparativa: Python vs hoja de cálculo

| Característica              | Python (script)                          | Hoja de cálculo                        |
|----------------------------|------------------------------------------|----------------------------------------|
| Separación datos/análisis  | ✅ Sí                                     | ❌ No                                   |
| Trazabilidad del análisis  | ✅ Alta (todo queda en el script)         | ❌ Baja (fórmulas ocultas en celdas)    |
| Reproducibilidad           | ✅ Fácil de compartir y replicar          | ❌ Difícil de seguir y documentar       |
| Creatividad en el análisis | ✅ Vistas dinámicas y personalizadas      | ❌ Limitada a lo que se ve en pantalla |

---

### 🧪 5. Ejemplo práctico en Python

```python
import pandas as pd

# Cargar datos desde un archivo CSV
datos = pd.read_csv("produccion.csv")

# Limpiar datos (eliminar nulos, corregir formatos)
datos = datos.dropna()
datos["fecha"] = pd.to_datetime(datos["fecha"])

# Agrupar por semana y calcular rendimiento medio
rendimiento_semanal = datos.groupby(pd.Grouper(key="fecha", freq="W"))["rendimiento"].mean()

# Mostrar resultados
print(rendimiento_semanal)
```
